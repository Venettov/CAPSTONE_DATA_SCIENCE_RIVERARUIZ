## 2. Consolidated, non-leaked feature engineering pipeline

This section is now the **single source of truth** for data preparation. It builds:

- lagged crime features
- lagged population-change features

### Preprocessing note

Disaster variables are often zero-inflated and right-skewed, so the notebook
still applies a **Yeo–Johnson power transformation** to shock variables, fit on
training data only and then transferred to the held-out test set.

For the SVI block, this revision also:
- replaces infinities before modeling,
- clips extreme values using train-only quantile caps, and
- reports diagnostics before PCA so warnings are easier to interpret.

```python
# --- STEP 1: INTEGRATE CRIME SHOCKS ---
def load_crime_data_fixed(file_path):
    xl = pd.ExcelFile(file_path)
    all_years = []
    for sheet_name in xl.sheet_names:
        if not sheet_name.isdigit():
            continue
        df_year = xl.parse(sheet_name)
        df_year.columns = [str(c).strip().lower() for c in df_year.columns]

        mun_cols = [c for c in df_year.columns if 'mun' in c]
        if mun_cols:
            df_year = df_year.rename(columns={mun_cols[0]: 'municipio'})

        df_year = df_year.loc[:, ~df_year.columns.duplicated()]
        df_year = df_year.rename(columns={'total ': 'total'})

        numeric_cols = ['murder', 'robbery', 'burglary', 'larceny / theft', 'total']
        for col in numeric_cols:
            if col in df_year.columns:
                df_year[col] = (
                    df_year[col]
                    .astype(str)
                    .str.replace('.', '', regex=False)
                    .str.replace(',', '', regex=False)
                )
                df_year[col] = pd.to_numeric(df_year[col], errors='coerce').fillna(0).astype(int)

        df_year['year'] = int(sheet_name)
        keep_cols = ['municipio', 'year'] + [c for c in numeric_cols if c in df_year.columns]
        all_years.append(df_year[keep_cols])

    return pd.concat(all_years, ignore_index=True)

crime_path = '../data/crime/PR_Crime_Summary_2010_2025.xlsx'
crime_df = load_crime_data_fixed(crime_path)

df['mun_key'] = df['municipio'].astype(str).str.upper().str.strip().str.replace(' MUNICIPIO', '', regex=False)
crime_df['mun_key'] = crime_df['municipio'].astype(str).str.upper().str.strip()

df = df.merge(crime_df.drop(columns=['municipio']), on=['mun_key', 'year'], how='left')
df = df.fillna(0)
df = df.drop(columns=['mun_key'])

# --- STEP 2: CORE RATES, TARGET, AND LAGS ---
df['murder_rate'] = (df['murder'] / df['total_population']) * 1000
df['robbery_rate'] = (df['robbery'] / df['total_population']) * 1000
df['burglary_rate'] = (df['burglary'] / df['total_population']) * 1000

df = df.sort_values(['municipio', 'year']).copy()
df['year'] = pd.to_numeric(df['year'], errors='coerce')
df['target_pop_change'] = df.groupby('municipio')['total_population'].pct_change() * 100

df['lag_violent_crime_rate'] = (
    df.groupby('municipio')['murder_rate'].shift(1) +
    df.groupby('municipio')['robbery_rate'].shift(1)
)
df['lag_property_crime_rate'] = df.groupby('municipio')['burglary_rate'].shift(1)

df['lag_pop_change_1'] = df.groupby('municipio')['target_pop_change'].shift(1)
df['lag_pop_change_2'] = df.groupby('municipio')['target_pop_change'].shift(2)
df['income_growth'] = df.groupby('municipio')['median_income_real'].pct_change() * 100

# --- STEP 3: SVI INPUTS ---
svi_cols = [
    'poverty_rate_pct',
    'unemployment_rate_pct',
    'per_capita_income',
    'no_hs_diploma_pct',
    'under_18_pct',
    'over_65_pct',
    'disability_pct',
    'single_parent_pct',
    'minority_pct',
    'limited_english_pct',
    'multi_unit_housing_pct',
    'mobile_homes_pct',
    'crowding_pct',
    'no_vehicle_pct',
    'pct_group_quarters'
]

for col in svi_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df['per_capita_income_inv'] = -df['per_capita_income']

SVI_PCA_COLS = [
    'poverty_rate_pct',
    'unemployment_rate_pct',
    'per_capita_income_inv',
    'no_hs_diploma_pct',
    'under_18_pct',
    'over_65_pct',
    'disability_pct',
    'single_parent_pct',
    'minority_pct',
    'limited_english_pct',
    'multi_unit_housing_pct',
    'mobile_homes_pct',
    'crowding_pct',
    'no_vehicle_pct',
    'pct_group_quarters'
]

BASE_NUMERIC_COLS = [
    'median_income_real',
    'max_wind_knots',
    'max_seismic_mag',
    'lag_pop_change_1',
    'lag_pop_change_2',
    'income_growth',
    'lag_violent_crime_rate',
    'lag_property_crime_rate',
    'lat',
    'lon'
]

for col in BASE_NUMERIC_COLS:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

df = df.replace([np.inf, -np.inf], np.nan)

MODEL_REQUIRED_COLS = ['municipio', 'year', 'target_pop_change'] + SVI_PCA_COLS + [c for c in BASE_NUMERIC_COLS if c in df.columns]
df_clean = df.dropna(subset=MODEL_REQUIRED_COLS).copy()

# --- STEP 4: CONSISTENT TRAIN/TEST SPLIT ---
TRAIN_END_YEAR = 2020
TEST_START_YEAR = 2021

train_mask = df_clean['year'] <= TRAIN_END_YEAR
test_mask = df_clean['year'] >= TEST_START_YEAR

if not train_mask.any() or not test_mask.any():
    raise ValueError("Time-based split produced an empty train or test set. Check the year range in df_clean.")

df_clean['split'] = np.where(train_mask, 'train', np.where(test_mask, 'test', 'unused'))
df_model = df_clean[df_clean['split'].isin(['train', 'test'])].copy()

train_df = df_model[df_model['split'] == 'train'].copy().reset_index(drop=True)
test_df = df_model[df_model['split'] == 'test'].copy().reset_index(drop=True)

# --- STEP 5: TRAIN-ONLY OUTLIER CAPPING FOR SVI INPUTS ---
svi_clip_bounds = {}
for col in SVI_PCA_COLS:
    lower = train_df[col].quantile(0.01)
    upper = train_df[col].quantile(0.99)
    svi_clip_bounds[col] = (lower, upper)
    train_df[col] = train_df[col].clip(lower=lower, upper=upper)
    test_df[col] = test_df[col].clip(lower=lower, upper=upper)

# --- STEP 6: TRAIN-ONLY PCA FOR VULNERABILITY INDEX ---
scaler_svi = StandardScaler()
pca_svi = PCA(n_components=1, random_state=42)

train_svi_scaled = scaler_svi.fit_transform(train_df[SVI_PCA_COLS])
test_svi_scaled = scaler_svi.transform(test_df[SVI_PCA_COLS])

train_df['vulnerability_index'] = pca_svi.fit_transform(train_svi_scaled).ravel()
test_df['vulnerability_index'] = pca_svi.transform(test_svi_scaled).ravel()

loading_series = pd.Series(pca_svi.components_[0], index=SVI_PCA_COLS)
orientation_cols = [c for c in ['poverty_rate_pct', 'unemployment_rate_pct', 'no_vehicle_pct'] if c in loading_series.index]
if orientation_cols and loading_series[orientation_cols].mean() < 0:
    train_df['vulnerability_index'] = -train_df['vulnerability_index']
    test_df['vulnerability_index'] = -test_df['vulnerability_index']
    loading_series = -loading_series

svi_minmax = MinMaxScaler()
train_df['vulnerability_index'] = svi_minmax.fit_transform(train_df[['vulnerability_index']])
test_df['vulnerability_index'] = svi_minmax.transform(test_df[['vulnerability_index']])

# --- STEP 7: INTERACTIONS AND FINAL FEATURE LISTS ---
train_df['svi_x_wind'] = train_df['vulnerability_index'] * train_df['max_wind_knots']
train_df['svi_x_quake'] = train_df['vulnerability_index'] * train_df['max_seismic_mag']
test_df['svi_x_wind'] = test_df['vulnerability_index'] * test_df['max_wind_knots']
test_df['svi_x_quake'] = test_df['vulnerability_index'] * test_df['max_seismic_mag']

FINAL_FEATURES = [
    'vulnerability_index',
    'median_income_real',
    'max_wind_knots',
    'max_seismic_mag',
    'lag_violent_crime_rate',
    'lag_property_crime_rate',
    'lag_pop_change_1',
    'lag_pop_change_2',
    'income_growth',
    'svi_x_wind',
    'svi_x_quake'
]

BASELINE_FEATURES = [
    'vulnerability_index',
    'median_income_real',
    'max_wind_knots',
    'max_seismic_mag',
    'lag_violent_crime_rate',
    'lag_property_crime_rate'
]

SHOCK_COLS = [c for c in [
    'max_wind_knots',
    'max_seismic_mag'
] if c in FINAL_FEATURES]

# Fit Yeo-Johnson on training data only and transfer to test data.
if SHOCK_COLS:
    power_transformer = PowerTransformer(method='yeo-johnson', standardize=False)
    train_df.loc[:, SHOCK_COLS] = power_transformer.fit_transform(train_df[SHOCK_COLS])
    test_df.loc[:, SHOCK_COLS] = power_transformer.transform(test_df[SHOCK_COLS])

    # Refresh interaction features after transforming the underlying shock variables.
    train_df['svi_x_wind'] = train_df['vulnerability_index'] * train_df['max_wind_knots']
    train_df['svi_x_quake'] = train_df['vulnerability_index'] * train_df['max_seismic_mag']
    test_df['svi_x_wind'] = test_df['vulnerability_index'] * test_df['max_wind_knots']
    test_df['svi_x_quake'] = test_df['vulnerability_index'] * test_df['max_seismic_mag']

# Canonical modeling matrices
X_train = train_df[FINAL_FEATURES].copy()
X_test = test_df[FINAL_FEATURES].copy()
y_train = train_df['target_pop_change'].copy()
y_test = test_df['target_pop_change'].copy()

X_train_baseline = train_df[BASELINE_FEATURES].copy()
X_test_baseline = test_df[BASELINE_FEATURES].copy()

# Year-block expanding-window validation folds
train_years_sorted = sorted(train_df['year'].dropna().unique())
cv_splits = []
for year in train_years_sorted[3:]:
    train_idx = train_df.index[train_df['year'] < year].to_numpy()
    val_idx = train_df.index[train_df['year'] == year].to_numpy()
    if len(train_idx) > 0 and len(val_idx) > 0:
        cv_splits.append((train_idx, val_idx))

if len(cv_splits) < 2:
    raise ValueError("Not enough yearly blocks to create expanding-window CV splits.")

cv_fold_summary = pd.DataFrame([
    {
        'fold': i + 1,
        'train_years': f"{int(train_df.iloc[tr]['year'].min())}-{int(train_df.iloc[tr]['year'].max())}",
        'validation_year': int(train_df.iloc[val]['year'].iloc[0]),
        'train_rows': len(tr),
        'validation_rows': len(val)
    }
    for i, (tr, val) in enumerate(cv_splits)
])

print("--- Phase 2 data preparation complete ---")
print(f"Training years: <= {TRAIN_END_YEAR}")
print(f"Testing years: >= {TEST_START_YEAR}")
print(f"Train rows: {len(train_df)}")
print(f"Test rows: {len(test_df)}")
print(f"Explained variance of PCA component: {pca_svi.explained_variance_ratio_[0]:.4f}")
print("\nFinal Phase 2 features:")
for f in FINAL_FEATURES:
    print(f" - {f}")

print("\nReduced-feature linear benchmark features:")
for f in BASELINE_FEATURES:
    print(f" - {f}")

print("\nExpanding-window year-block CV folds:")
display(cv_fold_summary)
```

## 3. Model comparison with year-block validation

This section now compares four categories of models on the **same target, same split, and same metric scale**:

- a **reduced-feature linear benchmark** refit in this notebook
- two **naive forecasting benchmarks**
- **Random Forest**
- **HistGradientBoosting**

Hyperparameter tuning now uses **expanding-window validation by year**, which is easier to defend
for municipality-year forecasting than row-order splits.

Each validation fold trains on **all years before year _t_** and validates on **year _t_**.

```python
# --- Baselines and naive benchmarks on the exact same split ---

linear_baseline = LinearRegression()
linear_baseline.fit(X_train_baseline, y_train)
y_pred_linear_baseline = linear_baseline.predict(X_test_baseline)

linear_baseline_r2 = r2_score(y_test, y_pred_linear_baseline)
linear_baseline_mae = mean_absolute_error(y_test, y_pred_linear_baseline)
linear_baseline_rmse = np.sqrt(mean_squared_error(y_test, y_pred_linear_baseline))

# Naive benchmark 1: predict no change
y_pred_naive_zero = np.zeros(len(y_test))
naive_zero_r2 = r2_score(y_test, y_pred_naive_zero)
naive_zero_mae = mean_absolute_error(y_test, y_pred_naive_zero)
naive_zero_rmse = np.sqrt(mean_squared_error(y_test, y_pred_naive_zero))

# Naive benchmark 2: persistence from last year's observed change
y_pred_naive_persistence = test_df['lag_pop_change_1'].to_numpy()
naive_persistence_r2 = r2_score(y_test, y_pred_naive_persistence)
naive_persistence_mae = mean_absolute_error(y_test, y_pred_naive_persistence)
naive_persistence_rmse = np.sqrt(mean_squared_error(y_test, y_pred_naive_persistence))

rf_model = RandomForestRegressor(
    n_estimators=300,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)
rf_model.fit(X_train, y_train)
y_pred_rf_default = rf_model.predict(X_test)

rf_default_r2 = r2_score(y_test, y_pred_rf_default)
rf_default_mae = mean_absolute_error(y_test, y_pred_rf_default)
rf_default_rmse = np.sqrt(mean_squared_error(y_test, y_pred_rf_default))

baseline_results = pd.DataFrame({
    'Model': [
        'Linear benchmark (reduced feature set)',
        'Naive benchmark: zero change',
        'Naive benchmark: persistence',
        'Random Forest (default)'
    ],
    'R2': [
        linear_baseline_r2,
        naive_zero_r2,
        naive_persistence_r2,
        rf_default_r2
    ],
    'MAE': [
        linear_baseline_mae,
        naive_zero_mae,
        naive_persistence_mae,
        rf_default_mae
    ],
    'RMSE': [
        linear_baseline_rmse,
        naive_zero_rmse,
        naive_persistence_rmse,
        rf_default_rmse
    ]
})

print("--- Same-split linear and naive benchmarks ---")
display(baseline_results)
```

## Phase 2 Conclusion: Nonlinear Modeling, Target Design, and Predictability Limits

Phase 2 should be read as both a **nonlinear forecasting notebook** and a **target-design diagnostic**.

The original annual-target modeling results establish the core baseline: municipio-level population change in Puerto Rico is difficult to predict in a strict held-out setting, and much of the predictive signal is concentrated in short-run demographic momentum. This is why lagged population change remains central across the baseline model comparisons, and why the original best annual-target model is still useful as a benchmark for the rest of the project.

The extension section then tests whether weak forecasting performance reflects three issues: **target noise**, **structural breaks**, or **overly coarse disaster timing**.

The most important result from the extension is that the **best overall specification in the final Phase 2 notebook is the 3-year forward average target with Lasso**. This indicates that medium-run population change is more learnable than one-year municipio change, and that smoothing the target materially improves held-out performance relative to the original annual-target setup.

At the same time, the notebook's extension results suggest that the simple regime indicators (`post_maria`, `post_earthquakes`, `post_covid`) and the first-pass disaster timing features do **not** improve on that winning smoothed-target specification in their current form. In other words, the main gain comes from **target redesign**, not from the added regime or timing variables.

This result should still be interpreted carefully. The winning 3-year forward average target is evaluated on a **smaller effective test window** than the original annual-target baseline, because longer forward targets reduce the number of usable end-of-sample observations. That means the improvement is meaningful and encouraging, but it is not directly comparable to the original 2021–2024 annual-target evaluation without that caveat.

The final interpretation of Phase 2 is therefore:

- the **original 1-year target baseline** remains the project’s benchmark for annual prediction,
- the **3-year forward average target with Lasso** is now the strongest result in the merged final notebook,
- simple regime flags and first-pass disaster timing features did not add value beyond the smoothed target,
- and the central challenge in Puerto Rico municipio forecasting appears to be **target instability at the annual horizon** rather than a complete absence of structural signal.

This merged notebook therefore preserves the original baseline evidence, demonstrates that target design matters materially for prediction, and continues to export the artifacts needed by later phases in the project.

## Phase 2 Extension: Target Smoothing, Regime Indicators, and Disaster Timing

This extension adds three controlled experiment rounds while keeping the original Phase 2 workflow intact.

### Round 1: Alternative target definitions
The same model suite is evaluated on:
- the original 1-year target
- a 2-year forward average target
- a 3-year forward average target

### Round 2: Regime indicators
Using the best target from Round 1, the notebook adds:
- `post_maria`
- `post_earthquakes`
- `post_covid`

### Round 3: Disaster timing features
Using the best target + regime setup from Round 2, the notebook adds:
- rolling 3-year storm burden
- rolling 3-year seismic burden
- years since major storm
- years since major quake

These experiments are designed to test whether smoother targets, structural-break indicators, and timing-aware disaster exposure improve held-out prediction relative to the original specification.


```python
# --- Phase 2 Extension: setup and helper functions ---
from pathlib import Path
import json
import numpy as np
import pandas as pd

from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor

analysis_dir = Path(".")

phase2_ext_df = None
if "train_df" in globals() and "test_df" in globals():
    phase2_ext_df = pd.concat([train_df.copy(), test_df.copy()], ignore_index=True)
elif "df_model" in globals():
    phase2_ext_df = df_model.copy()
elif "df_clean" in globals():
    phase2_ext_df = df_clean.copy()
elif "df" in globals():
    phase2_ext_df = df.copy()

if phase2_ext_df is None:
    raise ValueError("Could not identify a dataframe from the current Phase 2 notebook state.")

required_base_cols = ["municipio", "year", "target_pop_change"]
missing_base_cols = [c for c in required_base_cols if c not in phase2_ext_df.columns]
if missing_base_cols:
    raise ValueError(f"Phase 2 extension is missing required columns: {missing_base_cols}")

phase2_ext_df = phase2_ext_df.sort_values(["municipio", "year"]).reset_index(drop=True)

if "FINAL_FEATURES" in globals():
    EXT_BASE_FEATURES = [c for c in FINAL_FEATURES if c in phase2_ext_df.columns]
elif "BASELINE_FEATURES" in globals():
    EXT_BASE_FEATURES = [c for c in BASELINE_FEATURES if c in phase2_ext_df.columns]
else:
    candidate_features = [
        "vulnerability_index",
        "median_income_real",
        "max_wind_knots",
        "max_seismic_mag",
        "lag_violent_crime_rate",
        "lag_property_crime_rate",
        "lag_pop_change_1",
        "lag_pop_change_2",
        "income_growth",
    ]
    EXT_BASE_FEATURES = [c for c in candidate_features if c in phase2_ext_df.columns]

if not EXT_BASE_FEATURES:
    raise ValueError("Could not determine a valid Phase 2 feature set for the extension.")

print("Extension dataframe shape:", phase2_ext_df.shape)
print("Base feature set used for extension:", EXT_BASE_FEATURES)


def add_forward_average_targets(df_in, source_col="target_pop_change"):
    df = df_in.copy().sort_values(["municipio", "year"])
    grp = df.groupby("municipio")[source_col]
    df["target_pop_change_1y"] = grp.shift(-1)
    df["target_pop_change_2y_avg"] = (grp.shift(-1) + grp.shift(-2)) / 2.0
    df["target_pop_change_3y_avg"] = (grp.shift(-1) + grp.shift(-2) + grp.shift(-3)) / 3.0
    return df


def add_regime_flags(df_in):
    df = df_in.copy()
    df["post_maria"] = (df["year"] >= 2018).astype(int)
    df["post_earthquakes"] = (df["year"] >= 2020).astype(int)
    df["post_covid"] = (df["year"] >= 2020).astype(int)
    return df


def years_since_event(series, threshold=0.0):
    years_since = []
    last_event_idx = None
    for i, val in enumerate(series):
        if pd.notna(val) and val > threshold:
            last_event_idx = i
            years_since.append(0.0)
        else:
            years_since.append(np.nan if last_event_idx is None else float(i - last_event_idx))
    return pd.Series(years_since, index=series.index)


def add_disaster_timing_features(df_in, storm_col="max_wind_knots", seismic_col="max_seismic_mag"):
    df = df_in.copy().sort_values(["municipio", "year"])
    if storm_col in df.columns:
        df["storm_burden_3y"] = (
            df.groupby("municipio")[storm_col]
              .transform(lambda s: s.fillna(0).rolling(window=3, min_periods=1).sum())
        )
        df["years_since_major_storm"] = (
            df.groupby("municipio")[storm_col]
              .transform(lambda s: years_since_event(s.fillna(0), threshold=0.0))
        )
    if seismic_col in df.columns:
        df["seismic_burden_3y"] = (
            df.groupby("municipio")[seismic_col]
              .transform(lambda s: s.fillna(0).rolling(window=3, min_periods=1).sum())
        )
        df["years_since_major_quake"] = (
            df.groupby("municipio")[seismic_col]
              .transform(lambda s: years_since_event(s.fillna(0), threshold=0.0))
        )
    return df


def safe_rmse(y_true, y_pred):
    return float(np.sqrt(mean_squared_error(y_true, y_pred)))


def get_time_split(df_in):
    df = df_in.copy()
    if "split" in df.columns:
        tr = df[df["split"] == "train"].copy()
        te = df[df["split"] == "test"].copy()
        if len(tr) > 0 and len(te) > 0:
            return tr, te
    if "TRAIN_END_YEAR" in globals():
        train_end = TRAIN_END_YEAR
    else:
        train_end = 2020
    if "TEST_START_YEAR" in globals():
        test_start = TEST_START_YEAR
    else:
        test_start = 2021
    tr = df[df["year"] <= train_end].copy()
    te = df[df["year"] >= test_start].copy()
    return tr, te


def build_model_suite(random_state=42):
    models = {
        "Ridge": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", Ridge(alpha=1.0))
        ]),
        "Lasso": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", Lasso(alpha=0.01, max_iter=20000))
        ]),
        "RandomForest": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", RandomForestRegressor(
                n_estimators=400,
                random_state=random_state,
                max_depth=None,
                min_samples_leaf=2,
                n_jobs=-1
            ))
        ]),
        "HistGradientBoosting": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", HistGradientBoostingRegressor(
                random_state=random_state,
                learning_rate=0.05,
                max_depth=4,
                max_iter=300
            ))
        ]),
    }
    return models


def evaluate_spec(df_in, target_col, feature_cols, round_name, spec_name, random_state=42):
    work = df_in.copy()
    work = work.dropna(subset=[target_col]).copy()
    train_part, test_part = get_time_split(work)
    if train_part.empty or test_part.empty:
        raise ValueError(f"Empty train/test split for target {target_col} in spec {spec_name}")

    X_train = train_part[feature_cols].copy()
    X_test = test_part[feature_cols].copy()
    y_train = train_part[target_col].copy()
    y_test = test_part[target_col].copy()

    results = []

    momentum_features = [c for c in ["lag_pop_change_1", "lag_pop_change_2"] if c in feature_cols]
    if momentum_features:
        momentum_model = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", Ridge(alpha=1.0))
        ])
        momentum_model.fit(train_part[momentum_features], y_train)
        pred = momentum_model.predict(test_part[momentum_features])
        results.append({
            "round": round_name,
            "specification_name": spec_name,
            "target_name": target_col,
            "feature_set_name": f"{spec_name}__momentum_only",
            "model_name": "MomentumOnly_Ridge",
            "n_features": len(momentum_features),
            "train_years": sorted(train_part["year"].unique().tolist()),
            "test_years": sorted(test_part["year"].unique().tolist()),
            "test_r2": float(r2_score(y_test, pred)),
            "test_rmse": safe_rmse(y_test, pred),
            "test_mae": float(mean_absolute_error(y_test, pred)),
        })

    zero_pred = np.zeros(len(y_test))
    results.append({
        "round": round_name,
        "specification_name": spec_name,
        "target_name": target_col,
        "feature_set_name": f"{spec_name}__naive_zero",
        "model_name": "NaiveZero",
        "n_features": 0,
        "train_years": sorted(train_part["year"].unique().tolist()),
        "test_years": sorted(test_part["year"].unique().tolist()),
        "test_r2": float(r2_score(y_test, zero_pred)),
        "test_rmse": safe_rmse(y_test, zero_pred),
        "test_mae": float(mean_absolute_error(y_test, zero_pred)),
    })

    for model_name, model in build_model_suite(random_state=random_state).items():
        model.fit(X_train, y_train)
        pred = model.predict(X_test)
        results.append({
            "round": round_name,
            "specification_name": spec_name,
            "target_name": target_col,
            "feature_set_name": spec_name,
            "model_name": model_name,
            "n_features": len(feature_cols),
            "train_years": sorted(train_part["year"].unique().tolist()),
            "test_years": sorted(test_part["year"].unique().tolist()),
            "test_r2": float(r2_score(y_test, pred)),
            "test_rmse": safe_rmse(y_test, pred),
            "test_mae": float(mean_absolute_error(y_test, pred)),
        })

    return pd.DataFrame(results)


phase2_ext_df = add_forward_average_targets(phase2_ext_df)
print("Created alternative target columns:",
      [c for c in ["target_pop_change_1y", "target_pop_change_2y_avg", "target_pop_change_3y_avg"] if c in phase2_ext_df.columns])
```

## Round 1: Compare 1-Year, 2-Year Average, and 3-Year Average Targets


```python
# --- Round 1: target redesign experiments ---
round1_results = []

round1_targets = [
    ("target_pop_change", "baseline_original_target"),
    ("target_pop_change_2y_avg", "forward_avg_2y"),
    ("target_pop_change_3y_avg", "forward_avg_3y"),
]

for target_col, spec_name in round1_targets:
    if target_col not in phase2_ext_df.columns:
        continue
    print(f"Running Round 1 spec: {spec_name} on target {target_col}")
    res = evaluate_spec(
        phase2_ext_df,
        target_col=target_col,
        feature_cols=EXT_BASE_FEATURES,
        round_name="Round1_TargetRedesign",
        spec_name=spec_name,
        random_state=42,
    )
    round1_results.append(res)

if not round1_results:
    raise ValueError("Round 1 produced no results.")

round1_results_df = pd.concat(round1_results, ignore_index=True)

display(
    round1_results_df
    .sort_values(["model_name", "test_r2"], ascending=[True, False])
    .reset_index(drop=True)
)

round1_best = (
    round1_results_df[~round1_results_df["model_name"].isin(["NaiveZero"])]
    .sort_values("test_r2", ascending=False)
    .iloc[0]
)

BEST_TARGET_COL = round1_best["target_name"]
BEST_ROUND1_SPEC = round1_best["specification_name"]

print("Best Round 1 target:", BEST_TARGET_COL)
print("Best Round 1 specification:", BEST_ROUND1_SPEC)
display(round1_best.to_frame().T)
```

## Round 2: Add Regime Indicators to the Best Target


```python
# --- Round 2: regime indicator experiments ---
phase2_ext_regime_df = add_regime_flags(phase2_ext_df)

regime_flag_features = [c for c in ["post_maria", "post_earthquakes", "post_covid"] if c in phase2_ext_regime_df.columns]

round2_feature_sets = {
    "besttarget_baseline_features": EXT_BASE_FEATURES,
    "besttarget_plus_regime_flags": list(dict.fromkeys(EXT_BASE_FEATURES + regime_flag_features)),
}

interaction_features = []
if "max_wind_knots" in phase2_ext_regime_df.columns and "post_maria" in phase2_ext_regime_df.columns:
    phase2_ext_regime_df["max_wind_knots_x_post_maria"] = (
        phase2_ext_regime_df["max_wind_knots"].fillna(0) * phase2_ext_regime_df["post_maria"]
    )
    interaction_features.append("max_wind_knots_x_post_maria")

if "vulnerability_index" in phase2_ext_regime_df.columns and "post_maria" in phase2_ext_regime_df.columns:
    phase2_ext_regime_df["vulnerability_index_x_post_maria"] = (
        phase2_ext_regime_df["vulnerability_index"].fillna(0) * phase2_ext_regime_df["post_maria"]
    )
    interaction_features.append("vulnerability_index_x_post_maria")

if "max_seismic_mag" in phase2_ext_regime_df.columns and "post_earthquakes" in phase2_ext_regime_df.columns:
    phase2_ext_regime_df["max_seismic_mag_x_post_earthquakes"] = (
        phase2_ext_regime_df["max_seismic_mag"].fillna(0) * phase2_ext_regime_df["post_earthquakes"]
    )
    interaction_features.append("max_seismic_mag_x_post_earthquakes")

round2_feature_sets["besttarget_plus_regimes_and_interactions"] = list(
    dict.fromkeys(EXT_BASE_FEATURES + regime_flag_features + interaction_features)
)

round2_results = []
for spec_name, feature_cols in round2_feature_sets.items():
    valid_feature_cols = [c for c in feature_cols if c in phase2_ext_regime_df.columns]
    print(f"Running Round 2 spec: {spec_name} with {len(valid_feature_cols)} features")
    res = evaluate_spec(
        phase2_ext_regime_df,
        target_col=BEST_TARGET_COL,
        feature_cols=valid_feature_cols,
        round_name="Round2_RegimeIndicators",
        spec_name=spec_name,
        random_state=42,
    )
    round2_results.append(res)

round2_results_df = pd.concat(round2_results, ignore_index=True)
display(
    round2_results_df
    .sort_values(["model_name", "test_r2"], ascending=[True, False])
    .reset_index(drop=True)
)

round2_best = (
    round2_results_df[~round2_results_df["model_name"].isin(["NaiveZero"])]
    .sort_values("test_r2", ascending=False)
    .iloc[0]
)

BEST_ROUND2_SPEC = round2_best["specification_name"]
if BEST_ROUND2_SPEC in round2_feature_sets:
    BEST_ROUND2_FEATURES = [c for c in round2_feature_sets[BEST_ROUND2_SPEC] if c in phase2_ext_regime_df.columns]
else:
    BEST_ROUND2_FEATURES = EXT_BASE_FEATURES

print("Best Round 2 specification:", BEST_ROUND2_SPEC)
display(round2_best.to_frame().T)
```

## Round 3: Add Disaster Timing Features to the Best Target + Regime Setup


```python
# --- Round 3: disaster timing experiments ---
phase2_ext_timing_df = add_disaster_timing_features(phase2_ext_regime_df)

timing_features = [
    c for c in [
        "storm_burden_3y",
        "seismic_burden_3y",
        "years_since_major_storm",
        "years_since_major_quake",
    ]
    if c in phase2_ext_timing_df.columns
]

round3_feature_sets = {
    "besttarget_bestregime": BEST_ROUND2_FEATURES,
    "besttarget_bestregime_plus_disastertiming": list(dict.fromkeys(BEST_ROUND2_FEATURES + timing_features)),
}

round3_results = []
for spec_name, feature_cols in round3_feature_sets.items():
    valid_feature_cols = [c for c in feature_cols if c in phase2_ext_timing_df.columns]
    print(f"Running Round 3 spec: {spec_name} with {len(valid_feature_cols)} features")
    res = evaluate_spec(
        phase2_ext_timing_df,
        target_col=BEST_TARGET_COL,
        feature_cols=valid_feature_cols,
        round_name="Round3_DisasterTiming",
        spec_name=spec_name,
        random_state=42,
    )
    round3_results.append(res)

round3_results_df = pd.concat(round3_results, ignore_index=True)
display(
    round3_results_df
    .sort_values(["model_name", "test_r2"], ascending=[True, False])
    .reset_index(drop=True)
)

round3_best = (
    round3_results_df[~round3_results_df["model_name"].isin(["NaiveZero"])]
    .sort_values("test_r2", ascending=False)
    .iloc[0]
)

BEST_ROUND3_SPEC = round3_best["specification_name"]
print("Best Round 3 specification:", BEST_ROUND3_SPEC)
display(round3_best.to_frame().T)
```

## Extension Interpretation

Use these results to compare three possible explanations for weak held-out performance:

1. **Target noise** — if the 2-year or 3-year average target materially improves performance, then annual municipal change may be too noisy for stable prediction.
2. **Structural breaks** — if regime flags improve results, then Puerto Rico’s population process is likely not stable across the full 2010–2024 period.
3. **Disaster timing** — if rolling burden and recency features help, then migration and population adjustment respond more to cumulative or recent shocks than to simple annual maxima.

The main outcome to track is not just the single best model, but whether the forecasting problem becomes more stable, more interpretable, and less dominated by a narrow lag-only signal.


## Phase 2 Final Comparison: Baseline vs Extended Models

The intended interpretation is:

- The **baseline Phase 2 workflow** evaluates annual municipio-level population change using the original 1-year target and the original feature set.
